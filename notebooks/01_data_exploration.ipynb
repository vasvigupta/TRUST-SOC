# TRUST-SOC: 01 - Data Exploration & Statistical Audit
This notebook inspects the raw CICIDS2017 dataset distribution, features, missing values, and original labels.

```python
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add project root to sys.path
ROOT_DIR = Path.cwd().parent
if str(ROOT_DIR) not in sys.path:
    sys.path.append(str(ROOT_DIR))

from src.config.config import RAW_DIR, LABEL_MAPPING, PROJECT_CLASSES
from src.preprocessing.preprocess import load_cicids2017, inspect_labels
```

### 1. Load CICIDS2017 Raw Dataset

```python
df = load_cicids2017(RAW_DIR)
print(f"Dataset Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
df.head()
```

### 2. Inspect Original Labels & Class Distribution

```python
counts = inspect_labels(df)
plt.figure(figsize=(10, 5))
counts.head(10).plot(kind='barh', color='#1565C0')
plt.title('Top 10 CICIDS2017 Original Attack Labels')
plt.xlabel('Count')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

### 3. Check Missing & Infinite Values

```python
num_cols = df.select_dtypes(include=[np.number])
n_nan = num_cols.isnull().sum().sum()
n_inf = np.isinf(num_cols).sum().sum()
n_dup = df.duplicated().sum()

print(f"Missing (NaN) Values : {n_nan:,}")
print(f"Infinite (+/-Inf)     : {n_inf:,}")
print(f"Duplicate Rows        : {n_dup:,}")
```
