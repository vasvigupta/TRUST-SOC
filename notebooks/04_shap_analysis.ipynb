# TRUST-SOC: 04 - SHAP Explainability
This notebook demonstrates global feature attributions and per-prediction explanation vectors using SHAP TreeExplainer.

```python
import sys
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT_DIR = Path.cwd().parent
if str(ROOT_DIR) not in sys.path:
    sys.path.append(str(ROOT_DIR))

from src.detection.model import DetectionAgent
from src.explainability.shap_explainer import explain_single_prediction
from src.config.config import SHAP_DIR, PROCESSED_X_TEST
```

### 1. Inspect Top Global SHAP Features

```python
top_f_path = SHAP_DIR / 'top_features.json'
with open(top_f_path) as f:
    top_feats = json.load(f)

top_df = pd.DataFrame(top_feats)
display(top_df.head(10))

# Display plot
import IPython.display as ipd
ipd.Image(filename=str(SHAP_DIR / 'shap_summary_bar.png'))
```

### 2. Explain Individual Network Flow Record

```python
agent = DetectionAgent.load()
X_test = pd.read_parquet(PROCESSED_X_TEST)

sample_row = X_test.iloc[0]
explanation = explain_single_prediction(agent, sample_row, list(agent.encoder.classes_))

print(f"Predicted Class : {explanation['prediction']}")
print(f"Confidence Score: {explanation['confidence']:.4f}")
print("\nTop Contributing Features:")
for feat in explanation['top_features'][:5]:
    print(f"  {feat['feature']:<35} : {feat['shap_value']:+.4f}")
```
